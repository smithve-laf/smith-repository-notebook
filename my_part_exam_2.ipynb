```python
#importing modules
import numpy as np
import matplotlib.pyplot as plt
```

```python
#1. importing data
url1 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_neuron1.csv'
url2 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_neuron2.csv'

neuron1 = np.loadtxt(url1, delimiter = ',')
neuron2 = np.loadtxt(url2, delimiter = ',')
```

```python
#2. these spikes were recorded during a 30 second period
#preceding the first trial of the task
```

```python
#4. calculating interspike interval
ISI_neuron1 = np.diff(neuron1)
ISI_neuron2 = np.diff(neuron2)

#finding firing rate
print(len(neuron1))
print(len(neuron2))

firing_rate_neuron1 = 45/30
firing_rate_neuron2 = 123/30
print(firing_rate_neuron1)
print(firing_rate_neuron2)
```

```python
#3. 4 subplots
#1st plot
fig, ax = plt.subplots(2,2, figsize = (15,10))
ax[0,0].plot(neuron1, np.ones_like(neuron1), '.')
ax[0,0].plot(neuron2, np.ones_like(neuron2), '.')
ax[0,0].set_xlim([0,30]) #whole 30 seconds
ax[0,0].set_ylim([0,2])
ax[0,0].set_xlabel('Time [s]')
ax[0,0].set_title('Neuron 1 and Neuron 2')
ax[0,0].legend({'Neuron1', 'Neuron 2'})

#2nd and 3rd plot ISI
bins = np.arange(0, 0.5, 0.001)
ax[0,1].hist(ISI_neuron1, bins)
ax[0,1].set_xlabel('ISI [s]')
ax[0,1].set_ylabel('Counts')
ax[0,1].set_title('Neuron 1 ISI')

ax[1,0].hist(ISI_neuron2, bins)
ax[1,0].set_xlabel('ISI [s]')
ax[1,0].set_ylabel('Counts')
ax[1,0].set_title('Neuron 2 ISI')

#4th subplot FR
time_bins = np.arange(0, 30, 1)
fr_neuron1, _ = np.histogram(neuron1, time_bins)
fr_neuron2, _ = np.histogram(neuron2, time_bins)

ax[1,1].plot(time_bins[:-1], fr_neuron1, color = 'blue')
ax[1,1].plot(time_bins[:-1], fr_neuron2, color = 'red')
ax[1,1].set_xlabel('Time [s]')
ax[1,1].set_ylabel('Firing Rate')
ax[1,1].set_title('Neuron 1 and Neuron 2 FR')
ax[1,1].legend({'Neuron 1', 'Neuron 2'})
```
