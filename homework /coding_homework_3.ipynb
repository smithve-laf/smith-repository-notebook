```python
#step 2 importing
import numpy as np
from matplotlib import pyplot as plt
import scipy.stats as stats
```

```python
#step 3 importing data
url1 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv'
url2 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv'

v_rest_sst = np.loadtxt(url1)
v_rest_pvalb = np.loadtxt(url2)
```

```python
#step 4
#a find the means of both data
mean_sst = np.mean(v_rest_sst)
print(mean_sst)
mean_pvalb = np.mean (v_rest_pvalb)
print(mean_pvalb)

#b perform ttest to find if means of data come from different distributions
statistic, p_value = stats.ttest_ind(v_rest_sst, v_rest_pvalb) #7.16 x 10^-9
print(statistic, p_value)

#c plot histograms using if statement
if p_value < 0.05:

  fig, ax = plt.subplots(1,2)

#hist and labels
  ax[0].hist(v_rest_sst)
  ax[0].set_title('SST')
  ax[0].set_xlabel('Resting Membrane Potential (mV)')
  ax[0].set_ylabel('Frequency')

  ax[1].hist(v_rest_pvalb)
  ax[1].set_title('PVALB')
  ax[1].set_xlabel('Resting Membrane Potential (mV)')
  ax[1].set_ylabel('Frequency')
```

```python
#step 5 define function
def vrestHist():
  """
  This is a function that will plot histograms of v_rest_sst and v_rest_pvalb if the p-value is < 0.05
  """
  if p_value < 0.05:
    fig, ax = plt.subplots(1,2)

    ax[0].hist(v_rest_sst)
    ax[0].set_title('SST')
    ax[0].set_xlabel('Resting Membrane Potential (mV)')
    ax[0].set_ylabel('Frequency')

    ax[1].hist(v_rest_pvalb)
    ax[1].set_title('PVALB')
    ax[1].set_xlabel('Resting Membrane Potential (mV)')
    ax[1].set_ylabel('Frequency')
```

```python
#step 6 call vrestHist()
vrestHist()
```

```python
#step 7
#a loop and scatter plot

fig, ax = plt.subplots(1,1)

#b loop
for i in range(len(v_rest_sst)):
  x = v_rest_sst[i]
  y = v_rest_pvalb[i]

#c assigning colors to datapoints
  if x > -60 and y > -60:
    color = 'red'
  elif x > -60 and y < -60:
    color = 'green'
  elif x < -60 and y < -60:
    color = 'blue'
  elif x < -60 and y > -60:
    color = 'purple'

  ax.scatter(x,y, color=color)

ax.set_xlabel('SST Resting Membrane Potential (mV)')
ax.set_ylabel('PVALB Resting Membrane Potential (mV)')
ax.set_title('Scatterplot of SST and PVALB')
```

```python
#step 8 defining vrestScatter
def vrestScatter():
  """
  This is a function that will create a scatterplot of SST neurons and Pvalb neurons based on datapoint value
  """
  fig, ax = plt.subplots(1,1)

#b loop
  for i in range(len(v_rest_sst)):
    x = v_rest_sst[i]
    y = v_rest_pvalb[i]

#c assigning colors to datapoints
    if x > -60 and y > -60:
      color = 'red'
    elif x > -60 and y < -60:
      color = 'green'
    elif x < -60 and y < -60:
      color = 'blue'
    elif x < -60 and y > -60:
      color = 'purple'

    ax.scatter(x,y, color=color)

  ax.set_xlabel('SST Resting Membrane Potential (mV)')
  ax.set_ylabel('PVALB Resting Membrane Potential (mV)')
  ax.set_title('Scatterplot of SST and PVALB')
```

```python
#checking if function works
vrestScatter()
```
