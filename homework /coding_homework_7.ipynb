```python
#step 2
#importing modules
import numpy as np
import matplotlib.pyplot as plt
```

```python
#step 3
#importing data

url1 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/acc_hits_hw5.csv'
url2 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/acc_fa_hw5.csv'

acc_hits = np.loadtxt(url1, delimiter = ',')
acc_fa = np.loadtxt(url2, delimiter = ',')
```

```python
#step 4
#checking shape
print(acc_hits.shape) #correct choices
print(acc_fa.shape) #incorrect choices (false alarms)

#row = trials, columns = samples (timepoints)
#when a mouse responds to a stimulus, we call it a hit. when it responds to the fake stimulus its called a false alarm (acc_fa)
```

```python
#step 6
#time variable
time = np.linspace(0, 4, 8001)
print(time.shape) #checking to make sure its the same length as acc_hits, acc_fa
```

```python
#step 7
#find sampling rate = samples/time
sampling_rate = 8001/4
print(sampling_rate)

#sampling rate is 2000 Hertz
```

```python
#step 8
#plotting first trial of each array in subplots. time on x
fig, ax = plt.subplots(2,1, figsize = (10,8)) #can add sharey=True to see same range

ax[0].plot(time, acc_hits[0])
ax[0].set_xlabel('Time [s]')
ax[0].set_ylabel('LFP Amplitude')
ax[0].set_title('First Trial, Correct Choices')

ax[1].plot(time, acc_fa[0])
ax[1].set_xlabel('Time [s]')
ax[1].set_ylabel('LFP Amplitude')
ax[1].set_title('First Trial, Incorrect Choices')
```

#####9. When the graphs are set to the same range, Correct choices seem to have much stronger responses (higher amplitude oscillations more regular), but when you get a closer look at the false alarms, you can see the oscillation better (but they are still more irregular and have lower amplitude).

```python
#step 10
#calculate and plot ERP for acc_hits
#erp is avaerage of responses across trials.

plt.plot(time, acc_hits.mean(0))
plt.xlabel('Time [s]')
plt.ylabel('LFP Amplitude') #MAKE SURE THIS IS RIGHT
plt.title("ERP of Correct Responses")

#make vertical and horizontal lines
plt.plot([2,2], [-50, 50], 'k')
plt.hlines(0, time[0], time[-1], 'k')
```

#####11. There is evidence of consistent activity in the LFP surrounding screen touch (vertical black bar). There is neural response after the screen touch. The response is not random.


```python
#step 12
#calculate and plot ERP for acc_fa and ERP for acc_hits in one plot. time on x
plt.plot(time, acc_hits.mean(0))
plt.plot(time, acc_fa.mean(0))
plt.xlabel('Time [s]')
plt.ylabel('LFP Amplitude') #MAKE SURE THIS IS RIGHT
plt.title("ERP of Correct and Incorrect Responses")

#make vertical and horizontal lines
plt.plot([2,2], [-50, 50], 'k')
plt.hlines(0, time[0], time[-1], 'k')
```

#####13. There are differences in the ERPs between hits and false alarms. False alarms have small amplitude and less structure around the screen touch at 2 seconds. The hits have large amplitude response around that 2 second mark.
