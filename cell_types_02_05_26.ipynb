#These two histograms are scaled differently. If the x-value is the ISI then most of the data is before 50 milliseconds whereas the SST is centered around 50 milliseconds and beyond. The x-axis represents ISI and y-axis is the counts.

Let's re-plot our histograms, and make the x-axis ranges the same. Let's also create x-axis and y-axis labels for our plots.

```python
# Let's re-make our histograms!

axes = cell_types['ISI'].hist(by = cell_types['Marker'])

axes[0].set_xlim(0,250)
axes[1].set_xlim(0,250)

# Enter appropriate labels for the x-axis and y-axis of the first histogram inside of the parentheses - remember to use '' when creating a string!
#this is how you label the y and x axis. but it only does it for the first graph because [0].

axes[0].set_ylabel('Counts')
axes[0].set_xlabel('ISI')

# Create code that will do the same thing for your second histogram!
#do the same thing but for [1] to label the second graph

axes[1].set_ylabel('Counts')
axes[1].set_xlabel('ISI')

#when they are set the same, the PVALB neurons are clearly the faster spiking neurons.
```
