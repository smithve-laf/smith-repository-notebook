```python
#task 2
import numpy as np
from matplotlib import pyplot as plt
import scipy.stats as stats
```

```python
#importing data
url1 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv'
url2 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv'

sst = np.loadtxt (url1)
pvalb = np.loadtxt (url2)
```

```python
#concatenate data
#making our list into an array - one dimension
sst1 = np.array(sst)
pvalb1 = np.array(pvalb)

#concatenating the arrays
combined = np.array([sst1,pvalb1])
print(combined.shape)
```

```python
#making three subplots

def examfunction():
  """
  This function creates 3 subplots looking at resting membrane potential for SST and PVALB neurons
  """

  fig, ax = plt.subplots(1,3,figsize=(15,5))

#subplot 1
  for i in range(len(combined)):
    x = sst1[i]
    y = pvalb1[i]

    if x < -60 and y < -60:
      color = 'red'
    else:
      color = 'blue'

    ax[0].scatter(x,y, color=color)
    ax[0].set_ylabel('PVALB Resting Membrane Potential (mV)')
    ax[0].set_xlabel('SST Resting Membrane Potential (mV)')

#subplot 2

    ax[1].hist(combined[0])
    ax[1].set_ylabel('Frequency')
    ax[1].set_xlabel('SST Resting Membrane Potential (mV)')

#subplot 3
    ax[2].hist(combined[1])
    ax[2].set_ylabel('Frequency')
    ax[2].set_xlabel('PVALB Resting Membrane Potential (mV)')

    print(stats.ttest_ind(sst1, pvalb1))

```

```python
#calls the function
examfunction()
```
